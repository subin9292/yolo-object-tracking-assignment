```python
import torch
from ultralytics import YOLO

model = YOLO('yolov8n.pt')

results = model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=50,
    imgsz=512,  # Roboflow에서 Resize 512x512로 설정했으므로 맞춤
    batch=16,
    device=0 if torch.cuda.is_available() else 'cpu'
)
```

```python
from ultralytics import YOLO

# 1. 방금 학습된 최적의 모델 가중치 불러오기
best_model = YOLO('runs/detect/train/weights/best.pt')

# 2. 5초 동영상에 대해 객체 감지 및 추적(track) 실행
track_results = best_model.track(
    source='test_video1.mp4',  # 작업 폴더에 넣은 5초 동영상 파일명
    save=True,                # 박스와 추적 ID가 그려진 결과 동영상 자동 저장
    conf=0.5,                 # 신뢰도 기준 (만약 박스가 잘 안 잡히면 0.3으로 낮춰서 재실행)
    tracker="bytetrack.yaml"  # 객체 추적 알고리즘
)
```

```python
# 결과 영상이 .avi로 저장된 경우에만 실행하여 .mp4로 변환
#%pip install moviepy
from moviepy import VideoFileClip

clip = VideoFileClip("runs/detect/track/test_video.avi")
clip.write_videofile("result_video.mp4", codec="libx264")
```
